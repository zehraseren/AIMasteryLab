  6. Kategorik sütunların category tipine dönüştürülmesi (string -> category)
  7. Dönüşüm sonrası veri tipinin kontrol edilmesi
  8. Güncellenmiş veri setinin kaydedilmesi

```python
import pandas as pd
```

```python
# 1. Veri setinin yüklenmesi
df = pd.read_csv("/content/drive/MyDrive/DataScience/e_ticaret_veri_seti_eksik_veriler_duzenlendi.csv")
```

```python
# 2. Veri tiplerinin incelenmesi
df.dtypes
```

```python
# 3. Sayısal, kategorik ve tarih sütunlarının belirlenmesi
sayisal_sutunlar = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
sayisal_sutunlar
```

```python
kategorik_sutunlar = df.select_dtypes(include=["object"]).columns.tolist()
kategorik_sutunlar
```

```python
# 4. Sipariş tarihi sütununun datetime tipine dönüştürülmesi (string -> datetime)
df["siparis_tarihi"] = pd.to_datetime(df["siparis_tarihi"])
df["siparis_tarihi"].dtype
```

```python
# 5. Tarih sütunundan yeni özellikler üretilmesi (Feature extraction)
# Zaman bazlı analizlerde yıl, ay, gün ve hafta günü bilgilerini üretir.
df["siparis_yili"] = df["siparis_tarihi"].dt.year
df["siparis_ayi"] = df["siparis_tarihi"].dt.month
df["siparis_gunu"] = df["siparis_tarihi"].dt.day
df["haftanin_gunu"] = df["siparis_tarihi"].dt.day_name()
df.head()
```

```python
# 6. Kategorik sütunların category tipine dönüştürülmesi (string -> category)
kategorik_donusum_sutunlari = [
    "sehir",
    "kategori",
    "urun_adi",
    "odeme_turu",
    "musteri_tipi",
    "haftanin_gunu"
]
```

```python
for sutun in kategorik_donusum_sutunlari:
  df[sutun] = df[sutun].astype("category")
```

```python
# 7. Dönüşüm sonrası veri tipinin kontrol edilmesi
df.dtypes
```

```python
# 8. Güncellenmiş veri setinin kaydedilmesi
df.to_csv("e_ticaret_veri_seti_veri_tipleri_duzenlendi.csv", index=False)
```
